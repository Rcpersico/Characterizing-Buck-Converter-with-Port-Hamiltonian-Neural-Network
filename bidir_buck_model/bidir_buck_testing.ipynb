```python
import numpy as np
import sys
sys.path.insert(0, 'src')
from generate_dataset import load_dataset, save_dataset

# Fix training set
trajs = load_dataset('data/plecs_bidir_train.npz')
for td in trajs:
    td['regime'] = float(np.mean(td['i_L'] < -0.05))

unidir = sum(1 for td in trajs if td['regime'] < 0.05)
bidir = sum(1 for td in trajs if td['regime'] >= 0.05)
regimes = [td['regime'] for td in trajs]
print(f"Training set:")
print(f"  Total: {len(trajs)}")
print(f"  Unidirectional (<5% negative): {unidir}")
print(f"  Bidirectional (>=5% negative): {bidir}")
print(f"  Regime range: [{min(regimes)*100:.1f}%, {max(regimes)*100:.1f}%]")
print(f"  Mean regime: {np.mean(regimes)*100:.1f}%")
save_dataset(trajs, 'data/plecs_bidir_train.npz')
print("  Saved.")

# Fix test set too if it exists
from pathlib import Path
if Path('data/plecs_bidir_test.npz').exists():
    test_trajs = load_dataset('data/plecs_bidir_test.npz')
    for td in test_trajs:
        td['regime'] = float(np.mean(td['i_L'] < -0.05))
    save_dataset(test_trajs, 'data/plecs_bidir_test.npz')
    print(f"Test set: {len(test_trajs)} trajectories fixed.")
```
