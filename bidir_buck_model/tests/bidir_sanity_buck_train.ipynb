```python
import sys
sys.path.insert(0, 'src')
import numpy as np
import torch
import matplotlib.pyplot as plt
from generate_dataset import load_dataset
from phnn_bidir import BidirBuckPHNN
from pathlib import Path

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
ckpt = torch.load('models/sanity_check.pt', map_location='cpu', weights_only=False)

model = BidirBuckPHNN(L=100e-6, C=100e-6,
                    initial_R_L=0.10, initial_R_C=0.05).to(device)
model.load_state_dict(ckpt['model_state_dict'])
model.eval()

trajs = load_dataset('data/plecs_bidir_train.npz')

# Compute regime on the fly and pick a DCM-dominant one
dcm_trajs = []
for i, td in enumerate(trajs):
    pct_neg = float(np.mean(td['i_L'] < -0.05))
    if pct_neg >= 0.05:
        dcm_trajs.append((i, td, pct_neg))

print(f"Found {len(dcm_trajs)} DCM-dominant trajectories")
# Pick one with high DCM content
dcm_trajs.sort(key=lambda x: x[2], reverse=True)
idx, td, regime = dcm_trajs[0]
print(f"Using traj {idx} with {regime*100:.1f}% reverse i_L")

def rk4_rollout(model, x0, t, u, device):
    states = np.zeros((len(t), 2), dtype=np.float32)
    states[0] = x0
    for k in range(len(t)-1):
        dt_k = float(t[k+1] - t[k])
        def f(x_np):
            xt = torch.tensor(x_np[None], dtype=torch.float32, device=device)
            ut = torch.tensor(u[k:k+1], dtype=torch.float32, device=device)
            with torch.no_grad():
                return model(xt, ut).cpu().numpy()[0]
        k1 = f(states[k])
        k2 = f(states[k] + 0.5*dt_k*k1)
        k3 = f(states[k] + 0.5*dt_k*k2)
        k4 = f(states[k] +     dt_k*k3)
        states[k+1] = states[k] + (dt_k/6)*(k1+2*k2+2*k3+k4)
    return states

x0 = np.array([td['i_L'][0], td['V_C'][0]], dtype=np.float32)
u  = np.stack([td['alpha']*td['V_in'], td['i_o']], axis=1).astype(np.float32)
pred = rk4_rollout(model, x0, td['t'], u, device)

rmse_iL = np.sqrt(np.mean((pred[:,0] - td['i_L'])**2))
rmse_VC = np.sqrt(np.mean((pred[:,1] - td['V_C'])**2))
print(f"RMSE i_L: {rmse_iL:.4f} A")
print(f"RMSE V_C: {rmse_VC:.4f} V")

fig, axes = plt.subplots(2, 1, figsize=(12, 7))
t_ms = td['t'] * 1e3
axes[0].plot(t_ms, td['i_L'], 'k--', lw=1.5, label='PLECS')
axes[0].plot(t_ms, pred[:,0], 'b-',  lw=1.5, label='DCM PHNN')
axes[0].axhline(0, color='gray', lw=0.5, alpha=0.5, ls=':')
axes[0].set_ylabel('i_L [A]')
axes[0].legend(); axes[0].grid(True, alpha=0.3)

axes[1].plot(t_ms, td['V_C'], 'k--', lw=1.5, label='PLECS')
axes[1].plot(t_ms, pred[:,1], 'r-',  lw=1.5, label='DCM PHNN')
axes[1].set_ylabel('V_C [V]')
axes[1].set_xlabel('Time [ms]')
axes[1].legend(); axes[1].grid(True, alpha=0.3)

fig.suptitle(f'DCM PHNN sanity check — traj {idx}, {regime*100:.1f}% reverse i_L')
fig.tight_layout()
plt.savefig('tests/bidir_sanity_traj.png', dpi=150)
print("Saved: tests/bidir_sanity_traj.png")
```
